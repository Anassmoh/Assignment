```python
# Your code starts here

x = np.linspace(0, 10, 500)
y1 = np.sqrt(4.8*x)
y2 = (x-2.17)**2

plt.figure(figsize=(8,5))
plt.plot(x,y1, color="g", label='$f(x)= \sqrt{4.8x}$')
plt.plot(x,y2, color="C8", label='$g(x)=(x-2,17)^2$')
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc="upper left")
plt.title("Overview: Both functions")
plt.grid(True)

fig, ax = plt.subplots(2, figsize=(8,10))

plt.sca(ax[0])
plt.plot(x,y1, color="g", label='$f(x)= \sqrt{4.8x}$')
plt.plot(x,y2, color="C8", label='$g(x)=(x-2,17)^2$')
plt.title("Intersection Point 1 (zoomed)")
plt.xlim(0.4,1.2)
plt.ylim(0,5)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc ="upper left")
plt.grid(True)

plt.sca(ax[1])
plt.plot(x,y1, color="g", label='$f(x)= \sqrt{4.8x}$')
plt.plot(x,y2, color="C8", label='$g(x)=(x-2,17)^2$')
plt.title("Intersection Point 2 (zoomed)")
plt.xlim(3.8,4.6)
plt.ylim(1.5,6.5)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```
