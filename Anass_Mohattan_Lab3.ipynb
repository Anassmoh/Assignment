### My solution

```python
x = np.arange(-10,10,0.01)
y = 0*(x<-1.6) + (1.6 + x)*((-1.6<=x)&(x<1.4)) + (4.4-x)*((1.4<=x)&(x<4.4)) + 0*(4.4<=x)

fig, ax = plt.subplots(figsize=(8,5))

plt.plot(x,y, color="C8", label="f(x)")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Problem 1: Piecewise Function")
plt.grid(True)
plt.legend(loc="upper left")
plt.show()
```
